## make_inference ## 

```python
def make_inference(model: str, prompt: str):
    api_request_json = {
        "model": model,
        "messages": [
            {"role": "user", 
            "content": prompt}
        ],
        "temperature": 0
    }

    response = llama.run(api_request_json)
    #print(json.dumps(response.json(), indent=2))
    #print(response.json()['choices'][0]['message']['content'])

    return response
```

#### Division en Chunks ####

Para lidiar con el problema de los tokens se opto por dividir en chunks y realizar inferencia multiples veces pero con N cuentas en lugar de la totalidad del nivel. Luego se juntan todas las respuestas y se vuelve a realizar inferencia con este set reducido para dar una respuesta final.

```python
# Import the islice function from the itertools module
from itertools import islice
# Define a function to yield chunks of a specified size from an iterable
def generate_chunks(iterable: list, size: int):
    # Create an iterator from the input iterable
    iterator = iter(iterable)
    
    # Loop over the iterator, taking the first element in each iteration
    for first in iterator:
        # Yield a list consisting of the first element and the next 'size-1' elements from the iterator
        yield [first] + list(islice(iterator, size - 1))
```

```python
import re

def parse_hierarchy_response(response: str) -> list[str]:
    pattern = 'ACTIVO >|PASIVO >|CAPITAL CONTABLE >|CUENTAS DE ORDEN >'
    response_path = []

    for line in response.split('\n'):
        match = re.search(pattern, line)
        if match:
            start = match.start()
            response_path.append(line[start:].rstrip())

    return response_path

def use_inference(account: str, level: int, model: str):

    # Get context
    db = SessionLocal()
    response = db.query(Cuenta).filter(Cuenta.cuenta_minima==True, Cuenta.nivel==level).all()
    chunks = list(generate_chunks(response, size=75))
    matches_path = []
    matches_db = []
    llm_responses = []

    for chunk in chunks:
        prompt_context = '\n'.join([f'Cuenta: {c.path}' for c in chunk])
        prompt = f'A que cuenta de tu contexto corresponde "{account}" en caso de corresponder a mas de una, devolver todas a las que corresponda. Si no corresponde a ninguna no devuelvas nada. \n\npara contexto:{prompt_context}'
        response = make_inference(model=model, prompt=prompt)
        llm_responses.append(response)
        response = response.json()['choices'][0]['message']['content']

        matches_path.extend(parse_hierarchy_response(response))

    # Use previous answers concatenated to ask again
    prompt_context = '\n'.join([f'Cuenta: {mp}' for mp in matches_path])
    prompt = f'A que cuenta de tu contexto corresponde "{account}" en caso de corresponder a mas de una, devolver todas a las que corresponda. Si no corresponde a ninguna no devuelvas nada. \n\npara contexto:{prompt_context}'
    response = make_inference(model=model, prompt=prompt)
    llm_responses.append(response)
    response = response.json()['choices'][0]['message']['content']
    final_matches_path = parse_hierarchy_response(response)

    for mp in final_matches_path:
        account_db = db.query(Cuenta).filter(Cuenta.path==mp).first()
        if account_db:
            matches_db.append(account_db)
    
    return matches_db, llm_responses
```

```python
import pandas as pd

def load_data_df(nodos, padre='', root=''):
    global data

    for account in nodos:
        id = account['id']
        nombre = account['nombre']
        nivel = account['nivel']
        ruta = account['path']
        hijos = account['hijos']

        if nombre in ('ACTIVO', 'PASIVO', 'CUENTAS DE ORDEN', 'CAPITAL CONTABLE'):
            root = nombre

        if len(hijos) > 0:
            load_data_df(hijos, padre=nombre, root=root)

        row = {"id": id, 'nombre': nombre, 'nivel': nivel, 'ruta': ruta, 'padre': padre, 'root': root}
        data.append(row)

def create_df_accounts():
    db = SessionLocal()
    cuentas = db.query(Cuenta).all()
    data = [{"id": c.id, 
             'nombre': c.nombre, 
             'nivel': c.nivel, 
             'ruta': c.path, 
             'padre': str(c.path_padre).split(' > ')[-1]} for c in cuentas]

    df = pd.DataFrame.from_records(data)
    df = df.set_index("id")
    df = df.sort_index()

    db.close()
    return df

def generate_prompt_Q1(account: str) -> str:

    df = create_df_accounts()
    prompt = f'''¿La cuenta de **{account}** corresponde a un ACTIVO, PASIVO, CAPITAL CONTABLE o CUENTAS DE ORDEN?

Solamente debes responder con las opciones que se te brindaron.'''

    df_n1 = df[df['nivel'] == 2].groupby('padre').agg(
            hijos=('nombre','; '.join)
            ).reset_index()

    context = ''
    for _, row in df_n1.iterrows():
        context += f'Las siguientes son cuentas de {row.padre}: {row.hijos}\n\n'

    prompt = context + prompt

    return prompt

def generate_prompt_Q2(account: str, response_account: str, level: int = 2) -> str:
    
    df = create_df_accounts()
    prompt = f'''¿A que cuenta de **{response_account}** corresponde **{account}**? 
    
Solamente debes responder con las opciones que se te brindaron en tu contexto.'''

    cuentas = df[(df['nivel'] == level) & (df['padre'] == response_account)]['nombre']
    cuentas_str = '; '.join(cuentas)
    context = f'\n\nPara contexto: Las siguientes son cuentas de **{response_account}**: {cuentas_str}'

    prompt = prompt + context

    return prompt

def inference_by_channeling(path: str):
    db = SessionLocal()
    Q1 = generate_prompt_Q1(path)
    response = make_inference(model=model2, prompt=Q1)
    response = response.json()['choices'][0]['message']['content']
    responses = [response]

    levels = path.count('>')

    for level in range(2, levels+2):
        prompt = generate_prompt_Q2(account=path, response_account=response, level=level)
        response = make_inference(model=model2, prompt=prompt)
        response = response.json()['choices'][0]['message']['content']
        responses.append(response)
    
    path = ' > '.join(responses)
    account_db = db.query(Cuenta).filter(Cuenta.path==path).first()

    db.close()
    return ([account_db], path, responses)
```
